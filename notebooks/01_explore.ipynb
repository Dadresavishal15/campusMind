# CampusMind – Exploratory Data Analysis

This notebook provides a quick exploration of the synthetic student feedback dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams['figure.figsize'] = (10, 5)

# Load dataset
df = pd.read_csv('../data/synthetic_feedback.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
# Basic description
df.describe(include='all')
```

```python
# Missing values
print('Missing values per column:')
print(df.isnull().sum())
```

## Category Distribution

```python
cat_counts = df['category'].value_counts()

fig, ax = plt.subplots()
cat_counts.plot(kind='bar', ax=ax, color='steelblue', edgecolor='white')
ax.set_title('Feedback Count by Category')
ax.set_xlabel('Category')
ax.set_ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print(cat_counts)
```

## Severity Distribution

```python
sev_counts = df['severity'].value_counts()
colors = {'high': '#e74c3c', 'medium': '#f39c12', 'low': '#2ecc71'}
bar_colors = [colors.get(s, 'gray') for s in sev_counts.index]

fig, ax = plt.subplots()
sev_counts.plot(kind='bar', ax=ax, color=bar_colors, edgecolor='white')
ax.set_title('Feedback Count by Severity')
ax.set_xlabel('Severity')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

## Sample Feedback Items

```python
# Show 5 random feedback items per category
for cat in df['category'].unique():
    print(f'\n=== {cat.upper()} ===')
    samples = df[df['category'] == cat]['feedback_text'].sample(min(3, len(df[df['category']==cat])), random_state=42)
    for i, s in enumerate(samples, 1):
        print(f'  [{i}] {s}')
```

## Department × Category Heatmap

```python
import numpy as np

pivot = df.groupby(['department', 'category']).size().unstack(fill_value=0)

fig, ax = plt.subplots(figsize=(12, 5))
im = ax.imshow(pivot.values, aspect='auto', cmap='YlOrRd')
ax.set_xticks(range(len(pivot.columns)))
ax.set_yticks(range(len(pivot.index)))
ax.set_xticklabels(pivot.columns, rotation=45, ha='right')
ax.set_yticklabels(pivot.index)
ax.set_title('Feedback Count Heatmap: Department × Category')
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()

pivot
```
